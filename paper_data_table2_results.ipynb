```python
import pandas as pd

def summarize_df(df, cols, label):
    grp = df.groupby('index')[cols]
    mean_idx = grp.mean()
    std_idx  = grp.std(ddof=1)
    overall_mean = mean_idx.mean()
    overall_std  = std_idx.mean()
    return pd.DataFrame([{
        'Condition':    label,
        'Jaccard':      f"{overall_mean[cols[0]]:.2f}±{overall_std[cols[0]]:.2f}",
        'MedCPT (cos)': f"{overall_mean[cols[1]]:.2f}±{overall_std[cols[1]]:.2f}",
        'Scincl (cos)': f"{overall_mean[cols[2]]:.2f}±{overall_std[cols[2]]:.2f}",
        'Correctness':  f"{overall_mean[cols[3]]:.2f}±{overall_std[cols[3]]:.2f}"
    }])

experiments = [
    (df_phi4_raw_exp,
     ['raw_phi4_ji', 'raw_phi4_similarity_score_medcpt_dot',
      'raw_phi4_similarity_score_scincl_dot', 'raw_score_phi4'],
     '(RAW) Phi-4'),
    (df_phi4_bl_exp,
     ['ji_baseline_phi4', 'baseline_similarity_score_medcpt_dot_phi4',
      'baseline_similarity_score_scincl_dot_phi4', 'score_bseline_phi4'],
     '(BL) Phi-4'),
    (df_phi4_fl_exp,
     ['ji_fl_phi4', 'phi4_fl_similarity_score_medcpt_dot',
      'phi4_fl_similarity_score_scincl_dot', 'score_fl_phi4'],
     '(FL) Phi-4'),

    (df_llama31_8B_raw_exp,
     ['ji_raw_llama31_8B', 'similarity_score_medcpt_dot_raw_llama31_8B',
      'similarity_score_scincl_dot_raw_llama31_8B', 'score_raw_llama31_8B'],
     '(RAW) Llama-3.1 8B'),
    (df_llama31_8B_bl_exp,
     ['llama31_8B_ji_bl', 'llama31_8B_bl_similarity_score_medcpt_dot',
      'llama31_8B_bl_similarity_score_scincl_dot', 'llama31_8B_score_bseline'],
     '(BL) Llama-3.1 8B'),
    (df_llama31_8B_fl_exp,
     ['llama31_8B_ji_fl', 'llama31_8B_fl_similarity_score_medcpt_dot',
      'llama31_8B_fl_similarity_score_scincl_dot', 'llama31_8B_score_fl'],
     '(FL) Llama-3.1 8B'),

    (df_llama33_70B_raw_exp,
     ['ji_raw_llama33_70B', 'similarity_score_medcpt_dot_raw_llama33_70B',
      'similarity_score_scincl_dot_raw_llama33_70B', 'score_raw_llama33_70B'],
     '(RAW) Llama-3.3 70B'),
    (df_llama33_70B_bl_exp,
     ['llama33_70B_ji_bl', 'llama33_70B_bl_bl_similarity_score_medcpt_dot',
      'llama33_70B_bl_similarity_score_scincl_dot', 'llama33_70B_score_bl'],
     '(BL) Llama-3.3 70B'),
    (df_llama33_70B_fl_exp,
     ['llama33_70B_ji_fl', 'llama33_70B_fl_similarity_score_medcpt_dot',
      'llama33_70B_fl_similarity_score_scincl_dot', 'llama33_70B_fl_score'],
     '(FL) Llama-3.3 70B'),
]

# Compute and concatenate all summaries
summaries = [summarize_df(df, cols, label) for df, cols, label in experiments]
all_summary = pd.concat(summaries, ignore_index=True)
all_summary
```
